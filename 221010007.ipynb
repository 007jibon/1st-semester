```python

# lab final ->>>>>>>>>
# STUDENT_ID : 221010007
# Md Shafiqul Islam
# Lab assignment 2
# Set : B





areas = ["WA", "NT", "SA", "Q", "NSW", "V", "T", "C"]

colors_available = {}
for a in areas:
    colors_available[a] = ["Red", "Green", "Blue"]

graph = {
    "WA": ["NT", "SA", "C"],
    "NT": ["WA", "SA", "Q", "C"],
    "SA": ["WA", "NT", "Q", "NSW", "V", "C"],
    "Q": ["NT", "SA", "NSW", "C"],
    "NSW": ["Q", "SA", "V"],
    "V": ["SA", "NSW"],
    "T": [],
    "C": ["WA", "NT", "SA", "Q"]
}

steps = {"backtracks": 0}


def is_valid_choice(region, colour, current_assignment):
    neighbours = graph.get(region, [])
    for nb in neighbours:
        if nb in current_assignment:
            if current_assignment[nb] == colour:
                return False
    return True


def choose_next_region(current_assignment):
    not_colored = []
    for r in areas:
        if r not in current_assignment:
            not_colored.append(r)

    chosen = None
    smallest_options = float("inf")

    for r in not_colored:
        possible = 0
        for c in colors_available[r]:
            if is_valid_choice(r, c, current_assignment):
                possible += 1

        if possible < smallest_options:
            smallest_options = possible
            chosen = r

    return chosen


def search(solution):
    if len(solution) == len(areas):
        return solution

    next_region = choose_next_region(solution)

    for colour in colors_available[next_region]:
        if is_valid_choice(next_region, colour, solution):

            print(next_region, "->", colour)
            solution[next_region] = colour

            answer = search(solution)
            if answer is not None:
                return answer

            print("Removing", next_region)
            solution.pop(next_region)
            steps["backtracks"] += 1

    return None


result = search({})

print("\nFinal Coloring:")
print(result)
print("Backtrack count:", steps["backtracks"])
```
